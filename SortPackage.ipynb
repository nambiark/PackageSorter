```python
def sortPackages(width, height, length, mass):
    # Calculate volume
    volume = width * height * length
    # Determine if the package is bulky or heavy
    is_bulky = volume >= 1_000_000 or width >= 150 or height >= 150 or length >= 150
    # Determine if the package is heavy
    is_heavy = mass >= 20


    # Determine the category of the package
    if is_bulky and is_heavy:
        return "REJECTED"
    if is_bulky or is_heavy:
        return "SPECIAL"
    return "STANDARD"
```

```python
#Test cases

tests = [
    # (width, height, length, mass, expected, description)

    # STANDARD: not bulky, not heavy
    (10,  10,  10,  5,  "STANDARD", "small package, light"),
    (99,  99,  99,  19, "STANDARD", "just under volume threshold, just under heavy"),
    (1,   1,   1,   1,  "STANDARD", "tiny package"),

    # SPECIAL: bulky only (volume >= 1,000,000)
    (100, 100, 100, 5,  "SPECIAL",  "volume exactly 1,000,000"),
    (200, 50,  100, 5,  "SPECIAL",  "volume > 1,000,000"),

    # SPECIAL: bulky only (dimension >= 150)
    (150, 1,   1,   5,  "SPECIAL",  "width exactly 150"),
    (1,   150, 1,   5,  "SPECIAL",  "height exactly 150"),
    (1,   1,   150, 5,  "SPECIAL",  "length exactly 150"),
    (200, 1,   1,   5,  "SPECIAL",  "width > 150"),

    # SPECIAL: heavy only (mass >= 20)
    (10,  10,  10,  20, "SPECIAL",  "mass exactly 20"),
    (10,  10,  10,  50, "SPECIAL",  "mass > 20"),

    # REJECTED: both bulky and heavy
    (100, 100, 100, 20, "REJECTED", "volume exactly 1,000,000 and mass exactly 20"),
    (150, 1,   1,   20, "REJECTED", "dimension >= 150 and mass >= 20"),
    (200, 200, 200, 50, "REJECTED", "very large and very heavy"),
]

passed = 0
failed = 0

for width, height, length, mass, expected, desc in tests:
    result = sortPackages(width, height, length, mass)
    status = "PASS" if result == expected else "FAIL"
    if status == "PASS":
        passed += 1
    else:
        failed += 1
    print(f"[{status}] {desc}")
    if status == "FAIL":
        print(f"       Expected: {expected}, Got: {result}")

print(f"\n{passed}/{passed + failed} tests passed")
```
